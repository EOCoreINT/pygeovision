```python
# Read the copernicus provider module
import pygeofetch, pathlib

pkg = pathlib.Path(pygeofetch.__file__).parent

# Find all files that mention copernicus or token or download
for f in sorted(pkg.rglob("*.py")):
    src = f.read_text(errors='replace')
    if any(kw in src.lower() for kw in ['copernicus', 'cdse', 'zipper', 'odata']):
        print(f"\n=== {f.relative_to(pkg)} ===")
        print(src[:4000])
        print()
```
