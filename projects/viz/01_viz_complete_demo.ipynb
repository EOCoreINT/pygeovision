# PyGeoVision Visualization Layer

Complete demo of all visualization tools: `Map`, `RasterViewer`, `VectorViewer`, `ChangeViewer`, `TimeSeriesViewer`.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('.').parent))
from pygeovision.viz import Map, RasterViewer, VectorViewer, ChangeViewer, TimeSeriesViewer
print("PyGeoVision viz layer loaded")
```

## Interactive Map

```python
# Interactive Leaflet map with raster + vector overlay
m = Map(center=(5.6, -0.2), zoom=11, basemap="satellite")
# m.add_raster("scene.tif", colormap="RdYlGn", band=0, opacity=0.8)
# m.add_vector("buildings.geojson", color="red", fill_opacity=0.3)
print(m)
print("Export: m.export('map.html')")
print("Show:   m.show()  # interactive in Jupyter")
print()

# Split view (before/after)
split = Map.split_view("before.tif", "after.tif")
print(split)
print("Export: split.export('comparison.html')")
```

## Summary

| Viewer | Key methods | Best for |
|---|---|---|
| `Map` | `add_raster`, `add_vector`, `export` | Web delivery, field teams |
| `RasterViewer` | `rgb`, `ndvi`, `histogram`, `profile` | Scene exploration, QA |
| `VectorViewer` | `style_by_attribute`, `filter`, `heatmap` | GIS analysis |
| `ChangeViewer` | `split`, `difference`, `statistics` | Before/after comparison |
| `TimeSeriesViewer` | `animate`, `trend`, `anomaly` | Temporal monitoring |
