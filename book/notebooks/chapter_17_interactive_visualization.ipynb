## RasterViewer

```python
import tempfile

# Demonstrate with synthetic data
import numpy as np
import rasterio
from rasterio.transform import from_bounds

from pygeovision.viz import RasterViewer

tmp = tempfile.mkdtemp()
data = np.random.uniform(0.1, 0.8, (6, 64, 64)).astype("float32")
prof = {"driver":"GTiff","dtype":"float32","count":6,"width":64,"height":64,
        "crs":rasterio.crs.CRS.from_epsg(4326),
        "transform":from_bounds(-0.3,5.5,-0.05,5.7,64,64)}
with rasterio.open(f"{tmp}/scene.tif","w",**prof) as dst:
    dst.write(data)

rv = RasterViewer(f"{tmp}/scene.tif")
rv.rgb(red=2, green=1, blue=0).show()
rv.ndvi(nir_band=3, red_band=2).show()
rv.histogram().show()
```
