## Compute all key indices

```python
import numpy as np

# Synthetic 6-band array (B02,B03,B04,B08,B8A,B11)
H, W  = 128, 128
arr   = np.random.uniform(0.05, 0.8, (6, H, W)).astype("float32")
blue, green, red, nir, re4, swir = arr

# NDVI
ndvi = (nir - red) / (nir + red + 1e-8)
# NDWI
ndwi = (green - nir) / (green + nir + 1e-8)
# NDBI
ndbi = (swir - nir) / (swir + nir + 1e-8)
# EVI
evi  = 2.5 * (nir - red) / (nir + 6*red - 7.5*blue + 1)

for name, idx in [("NDVI",ndvi),("NDWI",ndwi),("NDBI",ndbi),("EVI",evi)]:
    print(f"{name}: min={idx.min():.3f}  max={idx.max():.3f}  mean={idx.mean():.3f}")
```

## Index-based land cover

```python
# Simple 4-class classification
lc = np.zeros((H, W), dtype="uint8")
lc[ndvi > 0.3]                               = 1  # Vegetation
lc[(ndwi > 0) & (ndvi <= 0.3)]              = 2  # Water
lc[(ndbi > 0.1) & (ndvi <= 0.3)]            = 3  # Built-up
lc[lc == 0]                                  = 4  # Bare soil

for i, name in enumerate(["Vegetation","Water","Built-up","Bare soil"], 1):
    print(f"{name}: {(lc==i).mean()*100:.1f}%")
```
