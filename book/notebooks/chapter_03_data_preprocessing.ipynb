## Cloud masking

```python
import numpy as np, rasterio

# Sentinel-2 SCL classes to keep
SCL_CLEAR = [4, 5, 6]  # vegetation, not-vegetated, water

# Example: compute cloud-free coverage
scl = np.random.randint(0, 12, (512, 512), dtype="uint8")
clear_mask = np.isin(scl, SCL_CLEAR)
coverage   = clear_mask.mean() * 100
print(f"Cloud-free coverage: {coverage:.1f}%")
```

## Normalisation strategies

```python
arr = np.random.randint(0, 10000, (6, 256, 256), dtype="int16").astype("float32")

# Scale factor (Sentinel-2 standard)
arr_sf = arr / 10000.0

# Percentile stretch
arr_pct = np.zeros_like(arr)
for b in range(arr.shape[0]):
    lo, hi = np.percentile(arr[b], 2), np.percentile(arr[b], 98)
    arr_pct[b] = np.clip((arr[b]-lo)/(hi-lo+1e-8), 0, 1)

print(f"Scale factor: {arr_sf.min():.3f} to {arr_sf.max():.3f}")
print(f"Percentile:   {arr_pct.min():.3f} to {arr_pct.max():.3f}")
```
