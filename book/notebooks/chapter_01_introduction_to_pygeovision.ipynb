# Chapter 1: Introduction to PyGeoVision

Quick-start tutorial covering the five-minute workflow.

```python
import pygeovision as pgv
print(f'PyGeoVision {pgv.__version__}')
```

## Define study area

```python
BBOX  = (-0.30, 5.50, -0.05, 5.70)   # Accra, Ghana
DATES = ("2026-06-01", "2026-06-30")
BANDS = ["B02","B03","B04","B08","B8A","B11"]
print(f"Bbox area: {(BBOX[2]-BBOX[0])*(BBOX[3]-BBOX[1])*111*111:.0f} km2 approx")
```

## Search and download

```python
client  = pgv.PyGeoVision()
results = client.search(bbox=BBOX, date_range=DATES, cloud_cover_max=20)
print(f"Found {len(results)} scenes")
for r in results[:3]:
    print(f"  {r.provider} {r.datetime[:10]} cloud={r.cloud_cover:.1f}%")
```

## Prepare for AI

```python
downloads = client.download(results[:1], output_dir="./data/intro/", bands=BANDS)
ready     = client.prepare_for_ai(downloads[0].path, model_type="foundation")
print(f"Shape: {ready['shape']}")
print(f"Steps: {ready['steps']}")
```

## Visualize

```python
from pygeovision.viz import RasterViewer
rv = RasterViewer(downloads[0].path)
rv.rgb(red=2, green=1, blue=0).show()
rv.ndvi(nir_band=3, red_band=2).export("./outputs/ndvi_accra.png")
```

## GeoAgent one-liner

```python
from pygeovision.agent import GeoAgent
agent = GeoAgent(client, output_dir="./results/intro/")
agent.set_context(bbox=BBOX, date="2026-06")
print("Agent ready. Call: agent.run('Map flood extent')")
```
