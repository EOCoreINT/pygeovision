# 01 · Getting Started with PyGeoVision

**PyGeoVision** combines two things normally kept in separate projects:
- 🛰️ **[PyGeoFetch](https://pypi.org/project/pygeofetch/)** — a real, independent PyPI dependency: ~24 satellite data provider integrations (search, download, radiometric preprocessing)
- 🤖 **A native AI layer** — implemented directly in PyGeoVision using pure PyTorch/HuggingFace Transformers/timm. PyGeoVision is independent of the separate `geoai` package -- it does not wrap or depend on it, despite an earlier version of this notebook claiming otherwise.

Real, verified numbers (confirmed directly against the installed code): **503** dataset registry entries, **98-121** model architectures (two separate, overlapping registries -- see the docs Architecture page), **51** AI task pipelines (all individually verified this audit cycle).

This notebook walks you through the first 30 minutes with PyGeoVision.

## 5 · First Search

Search for Sentinel-2 imagery over New York City with <15% cloud cover.

```python
# Search 22+ providers — open-access, no credentials needed
results = client.search(
    bbox=(-74.05, 40.70, -73.95, 40.80),   # Manhattan, NYC
    date_range=("2024-06-01", "2024-07-31"),
    providers=["planetary_computer", "aws_earth"],
    cloud_cover_max=15,
    max_results=10,
    sort_by="cloud_cover",
    use_cache=False,
)

print(f"Found {len(results)} scenes:\n")
for r in results[:5]:
    print(f"  {r}")
```

```python
# Inspect a SearchResult object
r = results[0]
print(f"ID:          {r.id}")
print(f"Provider:    {r.provider}")
print(f"Satellite:   {r.satellite}")
print(f"Date:        {r.date}")
print(f"Cloud cover: {r.cloud_cover:.1f}%")
print(f"Resolution:  {r.resolution_m}m")
print(f"BBox:        {r.bbox}")
print(f"Collection:  {r.collection}")
print(f"Assets:      {list(r.assets.keys())[:6]}")
print(f"Is SAR:      {r.is_sar}")
```

## 6 · First Download

```python

# Download the best 2 scenes with post-processing
downloads = client.download(
    results[:2],
    output_dir="./data/nyc/",
    parallel=4,
    verify_checksum=True,
    post_process=["reproject:EPSG:4326", "cog"],
)

print("\nDownload summary:")
for d in downloads:
    if d.success:
        print(f"  ✓ {d.scene_id[:50]}")
        print(f"    Path:  {d.path}")
        print(f"    Size:  {d.size_mb:.1f} MB")
        print(f"    Time:  {d.duration_seconds:.1f}s")
    else:
        print(f"  ✗ {d.scene_id}: {d.error}")
```
